```python
import numpy as np
import pandas as pd

r = 102303756

data = pd.read_csv("data.csv", encoding='latin1')
x = data["no2"].dropna().values

ar = 0.05 * (r % 7)
br = 0.3 * ((r % 5) + 1)

x = x / np.max(np.abs(x))
z = x + ar * np.arcsin(br * x)

mean = np.mean(z)
variance = np.var(z)

lambda_param = 1 / (2 * variance)
c = 1 / np.sqrt(2 * np.pi * variance)

print("mean =", mean)
print("lambda =", lambda_param)
print("c =", c)
```
